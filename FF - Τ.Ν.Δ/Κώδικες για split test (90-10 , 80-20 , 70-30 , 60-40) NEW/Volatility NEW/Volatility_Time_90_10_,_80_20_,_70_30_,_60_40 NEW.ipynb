```python
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.layers import Input, Dense, Embedding, Flatten, Concatenate
from tensorflow.keras.models import Model
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import mean_absolute_error
from tensorflow.keras.callbacks import EarlyStopping
from tabulate import tabulate

# Φόρτωση όλων των μετοχών
from google.colab import drive
drive.mount('/content/drive', force_remount=True)

symbols = ["AAPL","AIG", "BAC", "GS","MNST","MSFT","NKE","PFE","PLUG","SHOP","SPCE"]
path = "/content/drive/MyDrive/Βαση Δεδομενων/"

datasets = []

for symbol in symbols:
    df = pd.read_excel(f"{path}{symbol}_Data_Technical_Indicators_Combined.xlsx", parse_dates=[0])
    df.rename(columns={df.columns[0]: "Date"}, inplace=True)
    df.set_index("Date", inplace=True)
    df = df.sort_index()
    # Δημιουργία Volatility
    df["Volatility"] = df.iloc[:, 8].shift(-1)
    df["Symbol"] = symbol
    datasets.append(df)

# Κοινές ημερομηνίες για όλες τις μετοχές
common_dates = datasets[0].index
for df in datasets[1:]:
    common_dates = common_dates.intersection(df.index)

# Φιλτράρουμε όλα τα dataframes για τις κοινές ημερομηνίες
for i in range(len(datasets)):
    datasets[i] = datasets[i].loc[common_dates]

# Ενοποίηση όλων των datasets
dataset = pd.concat(datasets, axis=0)
dataset = dataset.dropna()

le = LabelEncoder()
dataset["Symbol_enc"] = le.fit_transform(dataset["Symbol"])
num_symbols = dataset["Symbol_enc"].nunique()

# Features και target
X_num = dataset.iloc[: , [1, 2, 3, 9, 10, 11, 5, 16]]
X_sym = dataset["Symbol_enc"]
y = dataset['Volatility']
dates = dataset.index

valid_idx = y.notna()
X_num = X_num.loc[valid_idx]
X_sym = X_sym.loc[valid_idx]
y = np.log1p(y.loc[valid_idx])

# Ορισμός test sizes και περιγραφές
test_sizes = [0.1, 0.2, 0.3, 0.4]
descriptions = ["90-10", "80-20", "70-30", "60-40"]

all_results = []

for test_size, description in zip(test_sizes, descriptions):
    print(f"\nProcessing for Test Size = {test_size} ({description})...\n")
    results = []

    train_mae_histories = []
    val_mae_histories = []

    train_loss_histories = []
    val_loss_histories = []

    total_len = len(common_dates)
    total_test_len = int(total_len * test_size)
    print(f"Total length: {total_len}")
    print(f"Total test length: {total_test_len}")

    # Εκτέλεση 10 επαναλήψεων
    for i in range(10):
        # Επιλογή τυχαίου start index για το test set
        start_idx = np.random.randint(0, total_len - total_test_len + 1)
        end_idx = start_idx + total_test_len

        # Χρησιμοποιούμε κοινές ημερομηνίες για όλες τις μετοχές
        test_dates = common_dates[start_idx:end_idx]

        X_num_train = X_num.loc[~X_num.index.isin(test_dates)]
        X_sym_train = X_sym.loc[~X_sym.index.isin(test_dates)]
        y_train = y.loc[~y.index.isin(test_dates)]

        X_num_test = X_num.loc[X_num.index.isin(test_dates)]
        X_sym_test = X_sym.loc[X_sym.index.isin(test_dates)]
        y_test = y.loc[y.index.isin(test_dates)]

        val_size = 0.2
        train_dates_sorted = X_num_train.index.unique().sort_values()

        val_len = int(len(train_dates_sorted) * val_size)

        train_dates_final = train_dates_sorted[:-val_len]
        val_dates = train_dates_sorted[-val_len:]

        X_num_train_final = X_num_train.loc[X_num_train.index.isin(train_dates_final)]
        X_sym_train_final = X_sym_train.loc[X_sym_train.index.isin(train_dates_final)]
        y_train_final = y_train.loc[y_train.index.isin(train_dates_final)]

        X_num_val = X_num_train.loc[X_num_train.index.isin(val_dates)]
        X_sym_val = X_sym_train.loc[X_sym_train.index.isin(val_dates)]
        y_val = y_train.loc[y_train.index.isin(val_dates)]

        start_date = X_num_test.index[0].strftime("%Y-%m-%d")
        end_date = X_num_test.index[-1].strftime("%Y-%m-%d")
        print(f"Training for experiment = {i+1} | Test window: {start_date} → {end_date}")

        # Feature scaling
        sc = StandardScaler()
        X_num_train_final = sc.fit_transform(X_num_train_final)
        X_num_val = sc.transform(X_num_val)
        X_num_test = sc.transform(X_num_test)

        # Νευρωνικό Δίκτυο
        num_input = Input(shape=(X_num_train_final.shape[1],), name="num_input")
        sym_input = Input(shape=(1,), name="sym_input")

        embed_dim = 2
        sym_embed = Embedding(input_dim=num_symbols,output_dim=embed_dim,name="symbol_embedding")(sym_input)
        sym_embed = Flatten()(sym_embed)
        x = Concatenate()([num_input, sym_embed])
        x = Dense(128, activation="relu")(x)
        x = Dense(64, activation="relu")(x)
        x = Dense(32, activation="relu")(x)
        x = Dense(16, activation="relu")(x)
        output = Dense(1, activation="linear")(x)
        model = Model(inputs=[num_input, sym_input], outputs=output)
        model.compile(optimizer='adam', loss='mean_squared_error', metrics=['mae'])
        early_stopping = EarlyStopping(monitor="val_loss", patience=10, restore_best_weights=True)
        history=model.fit([X_num_train_final, X_sym_train_final],y_train_final,validation_data=([X_num_val, X_sym_val], y_val),batch_size = 16, epochs = 1000,callbacks=[early_stopping], verbose=0)

        train_mae_histories.append( history.history["mae"])
        val_mae_histories.append( history.history["val_mae"])

        train_loss_histories.append(history.history["loss"])
        val_loss_histories.append(history.history["val_loss"])

        y_pred = model.predict([X_num_test,X_sym_test])
        mae = mean_absolute_error(y_test, y_pred)

        print(f"Experiment {i+1} | Test period: {start_date} → {end_date} | MAE: {mae:.4f}")
        results.append([i + 1, f"{start_date} → {end_date}", mae])

    max_epochs = max(max(len(h) for h in train_mae_histories), max(len(h) for h in val_mae_histories))

    train_mae_array = np.full((len(train_mae_histories), max_epochs ), np.nan)

    val_mae_array = np.full((len(val_mae_histories), max_epochs ), np.nan)

    train_loss_array = np.full((len(train_loss_histories), max_epochs ), np.nan)

    val_loss_array = np.full((len(val_loss_histories), max_epochs ), np.nan)

    for j, h in enumerate(train_mae_histories):
      train_mae_array[j, :len(h)] = h

    for j, h in enumerate(val_mae_histories):
      val_mae_array[j, :len(h)] = h

    for j, h in enumerate(train_loss_histories):
      train_loss_array[j, :len(h)] = h

    for j, h in enumerate(val_loss_histories):
      val_loss_array[j, :len(h)] = h

    mean_train_mae = np.nanmean(train_mae_array, axis=0)
    mean_val_mae = np.nanmean(val_mae_array, axis=0)

    std_train_mae = np.nanstd(train_mae_array, axis=0)
    std_val_mae = np.nanstd(val_mae_array, axis=0)

    mean_train_loss = np.nanmean(train_loss_array, axis=0)
    mean_val_loss = np.nanmean(val_loss_array, axis=0)

    std_train_loss = np.nanstd(train_loss_array, axis=0)
    std_val_loss = np.nanstd(val_loss_array, axis=0)

    epochs = np.arange(1, max_epochs + 1)

    plt.figure(figsize=(10, 6))
    plt.plot(epochs, mean_train_mae, label="Mean Training MAE")
    plt.fill_between(epochs, mean_train_mae - std_train_mae, mean_train_mae + std_train_mae, alpha=0.2)
    plt.plot(epochs, mean_val_mae, label="Mean Validation MAE")
    plt.fill_between(epochs, mean_val_mae - std_val_mae, mean_val_mae + std_val_mae, alpha=0.2)
    plt.xlabel("Epoch")
    plt.ylabel("MAE")
    plt.title(f"Training and Validation MAE - " f"Split {description}")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(10, 6))
    plt.plot(epochs, mean_train_loss, label="Mean Training Loss")
    plt.fill_between(epochs, mean_train_loss - std_train_loss, mean_train_loss + std_train_loss, alpha=0.2)
    plt.plot(epochs, mean_val_loss, label="Mean Validation Loss")
    plt.fill_between(epochs, mean_val_loss - std_val_loss, mean_val_loss + std_val_loss, alpha=0.2)
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.title(f"Training and Validation Loss - " f"Split {description}")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    table = tabulate(results, headers=["Experiment", "Test Period", "MAE"], tablefmt="pretty")
    all_results.append(f"\nTrain - Test ---> {description}")
    all_results.append(table)

# Εκτύπωση τελικών αποτελεσμάτων
print("\nFinal Results for All Test Sizes:\n")
for result in all_results:
    print(result)
```
